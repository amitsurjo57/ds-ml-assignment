```python
import seaborn as sns
import matplotlib.pyplot as plt
```

**Do the following tasks:**
* set style to '`white`'
* load titanic dataset into '`titanic`', tips into '`tips`' and iris into '`iris`'

```python
titanic = sns.load_dataset('titanic')
```

```python
tips = sns.load_dataset('tips')
```

```python
iris = sns.load_dataset('iris')
```

**Recreate the plot below. Use tips dataset. Hint: pay attention to x and y labels.**

```python
boxplot_tips = sns.boxplot(data=tips, x='day', y='tip', hue='sex')
boxplot_tips
```

**Recreate the plot below. Use tips dataset. Hint: pay attention to x and y labels.**

```python
jointgrid_tips = sns.JointGrid(data=tips, x='total_bill', y='tip')
jointgrid_tips.plot(sns.scatterplot, sns.histplot)
jointgrid_tips
```

**Recreate the distplot given below. Use titanic dataset. Hint: pay attention to x and y labels.**

```python
displot_titanic = sns.displot(data=titanic, x='fare', bins=10)
displot_titanic
```

**Recreate the boxplot below using titanic dataset. Hint: pay attention to x and y labels.**

```python
boxplot_titanic = sns.boxplot(data=titanic, x='class', y='age', hue='alive')
boxplot_titanic
```

**Recreate the plot below using titanic dataset. Hint: pay attention to x and y labels.**

```python
jointgrid_titanic = sns.JointGrid(data=titanic, x='fare', y='age')
jointgrid_titanic.plot(sns.scatterplot, sns.histplot)
jointgrid_titanic
```

**Recreate the plot below. Use titanic dataset. **

```python
boxplot_titanic = sns.boxplot(data=titanic, x='class', y='age')
boxplot_titanic
```

**Can you plot how many male and female were on titanic? Separate them based on their class.**

```python
countplot_titanic = sns.countplot(data=titanic, x='sex', hue='class')
countplot_titanic
```

**Recreate the plot below. Use titanic dataset. Hint: Matrix plots**

```python
numerical_titanic = titanic.select_dtypes(include=['number'])

corr_titanic = numerical_titanic.corr()

heatmap_titanic = sns.heatmap(corr_titanic, annot=True, cmap='coolwarm', fmt=".2f")

heatmap_titanic.set_title('Correlation Matrix of Titanic Dataset')
```
