```python
class Car:
  def __init__(self, brand="Unknown", color = "white"):
    self.brand = brand
    self.color = color

  def display(self):
    print(f"Brand: {self.brand}")
    print(f"Color: {self.color}")


bmw = Car("BMW", "Black")

bmw.display()
```
