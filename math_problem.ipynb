```python
class MathOperation:
  def add(self, a , b, c = 0):
    print(f"Addition: {a+b+c}")

class AdvancedMath(MathOperation):
  def add(self, a , b, c = 1):
    print(f"Multiplication: {a*b*c}")


adm = AdvancedMath()

adm.add(12, 12)
```
