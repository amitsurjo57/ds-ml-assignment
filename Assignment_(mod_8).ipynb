```python
import pandas as pd

df = pd.read_csv("/content/sample_data/bollwood_dataset")
```

```python
length = len(df)
print(f"Here has {length} records")
display(df)
```

```python
genre_counts = df['Genre'].value_counts(ascending=False)
display(genre_counts)

most_released_genre = genre_counts.index[0]
highest_genre_release = genre_counts.iloc[0]

print(f"This genre {most_released_genre} is most released with highest count {highest_genre_release}")
```

```python
cross_teb = pd.crosstab(df['Genre'], df['ReleaseTime'])

display(cross_tab)
```

```python
df['ReleaseDate'] = pd.to_datetime(df['ReleaseDate'], format='%d-%b-%y')
df['Month'] = df['ReleaseDate'].dt.month
month_counts = df['Month'].value_counts()

most_common_month_num = month_counts.index[0]
most_common_month_count = month_counts.iloc[0]

# Get the month name from the month number
month_name = pd.to_datetime(most_common_month_num, format='%m').strftime('%B')

print(f"The month with the maximum number of movie releases is {month_name} with {most_common_month_count} releases.")
```

```python
high_budget_movies = df[df['Budget (₹ Crore)'] >= 25]

high_budget_month_counts = high_budget_movies['Month'].value_counts()

if not high_budget_month_counts.empty:
  most_common_high_budget_month_num = high_budget_month_counts.index[0]
  most_common_high_budget_month_count = high_budget_month_counts.iloc[0]

  month_name = pd.to_datetime(most_common_high_budget_month_num, format='%m').strftime('%B')

  print(f"The month with the maximum number of high-budget movie releases (budget >= 25 crore) is {month_name} with {most_common_high_budget_month_count} releases.")
else:
  print("There are no high-budget movies (budget >= 25 crore) in the dataset.")
```

```python
df['ROI'] = (df['BoxOfficeCollection (₹ Crore)'] - df['Budget (₹ Crore)']) / df['Budget (₹ Crore)']

top_10_roi_movies = df.sort_values(by='ROI', ascending=False).head(10)

print("Top 10 movies with maximum Return on Investment (ROI):")
display(top_10_roi_movies[['MovieName', 'Budget (₹ Crore)', 'BoxOfficeCollection (₹ Crore)', 'ROI']])
```

```python
average_roi_by_release_time = df.groupby('ReleaseTime')['ROI'].mean().sort_values(ascending=False)

print("Average ROI for different release times:")
display(average_roi_by_release_time)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.histplot(df['Budget (₹ Crore)'], kde=True, bins=10)
plt.title('Distribution of Movie Budgets')
plt.xlabel('Budget (₹ Crore)')
plt.ylabel('Frequency')
plt.show()
```

```python
comedy_movies = df[df['Genre'].str.contains('Comedy', case=False, na=False)]
drama_movies = df[df['Genre'].str.contains('Drama', case=False, na=False)]

plt.figure(figsize=(12, 6))
sns.histplot(comedy_movies['ROI'], kde=True, color='skyblue', label='Comedy', bins=10)
sns.histplot(drama_movies['ROI'], kde=True, color='lightcoral', label='Drama', bins=10)

plt.title('Distribution of ROI for Comedy vs Drama Movies')
plt.xlabel('ROI')
plt.ylabel('Frequency')
plt.legend()
plt.show()

print("\nAverage ROI for Comedy movies:")
display(comedy_movies['ROI'].mean())

print("\nAverage ROI for Drama movies:")
display(drama_movies['ROI'].mean())
```

```python
df['YoutubeLikes'] = df['YoutubeLikes'].str.replace(r'\(Trailer likes\)', '', regex=True).astype(int)

correlation = df['BoxOfficeCollection (₹ Crore)'].corr(df['YoutubeLikes'])

print(f"The correlation between Box Office Collection and YouTube Likes is: {correlation:.2f}")

if correlation > 0:
    print("The correlation is positive, meaning as YouTube likes increase, box office collection tends to increase.")
elif correlation < 0:
    print("The correlation is negative, meaning as YouTube likes increase, box office collection tends to decrease.")
else:
    print("There is no significant linear correlation between Box Office Collection and YouTube Likes.")
```

```python
plt.figure(figsize=(12, 6))
sns.boxplot(x='Genre', y='YoutubeLikes', data=df)
plt.title('Distribution of YouTube Likes by Genre')
plt.xlabel('Genre')
plt.ylabel('YouTube Likes')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
df['YoutubeViews'] = df['YoutubeViews'].str.replace(r'\(Trailer views\)', '', regex=True).astype(int)
df['YoutubeDislikes'] = df['YoutubeDislikes'].str.replace(r'\(Trailer dislikes\)', '', regex=True).astype(int)

numerical_cols = ['Budget (₹ Crore)', 'BoxOfficeCollection (₹ Crore)', 'YoutubeViews', 'YoutubeLikes', 'YoutubeDislikes']
correlation_matrix = df[numerical_cols].corr()

plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Matrix of Numerical Variables')
plt.show()
```
