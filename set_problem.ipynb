```python
set_1 = set(input("Enter set 1: ").split(' '))

set_2 = set(input("Enter set 2: ").split(' '))

print("Union: ", set_1.union(set_2))

print("Intersection: ", set_1.intersection(set_2))

print("Difference: ", set_1.difference(set_2))
```
